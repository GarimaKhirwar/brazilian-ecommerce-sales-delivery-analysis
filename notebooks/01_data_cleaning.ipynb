```python
order_reviews=pd.read_csv("../data/01_raw/olist_order_reviews_dataset.csv") 
```

```python
order_reviews.head()
```

```python
order_reviews.info()
```

```python
order_reviews.isna().sum()
```

```python
order_reviews["no_review"]=order_reviews["review_comment_message"].isnull()
```

```python
order_reviews.head(5)
```

```python
# Create the separator ONLY if both columns have data
separator = np.where(order_reviews['review_comment_title'].notnull() & order_reviews['review_comment_message'].notnull(), ' - ', '')
# Combine everything in one go
order_reviews['review_message'] = (order_reviews['review_comment_title'].fillna('') + separator + order_reviews['review_comment_message'].fillna('')).str.strip()
# now drop title and message columns
order_reviews = order_reviews.drop(columns=['review_comment_title', 'review_comment_message'])
```

```python
order_reviews.head()
```

```python
# rename the req columns 
order_reviews.rename(columns={"no_review":"no_review_message"},inplace=True)
```

```python
# changing the order
# Define the order
cols = [
    'review_id', 
    'order_id', 
    'review_score', 
    'no_review_message', 
    'review_message', 
    'review_creation_date', 
    'review_answer_timestamp'
]

# Apply the order
order_reviews = order_reviews[cols]
```

```python
cols = ['review_creation_date', 'review_answer_timestamp']
order_reviews[cols] = order_reviews[cols].apply(pd.to_datetime, format='%Y-%m-%d %H:%M:%S')
order_reviews[cols].dtypes
```

```python
products = pd.read_csv("../data/01_raw/olist_products_dataset.csv")
products.head()
```

```python
products.info()
```

```python
products.isna().sum()
```

```python
products.dropna(axis = 0, inplace=True)
```

```python
products.isna().sum()
```

```python
cols = ['product_name_lenght', 'product_description_lenght', 'product_photos_qty']
products[cols] = products[cols].astype(int)
products[cols].dtypes
```

```python
cat_translations = pd.read_csv("../data/01_raw/product_category_name_translation.csv")
cat_translations.head() 
```

```python
# translating the category names to English
# 1. Create the translation dictionary
# Setting the Portuguese name as index and English as the value
trans_dict = cat_translations.set_index("product_category_name")["product_category_name_english"].to_dict()
# 2. Apply it to your main orders DataFrame
products['product_category_ename'] = products['product_category_name'].map(trans_dict)
products.drop(columns=['product_category_name'], inplace=True)
products.head()
```

```python
cols = ['product_id', 'product_category_ename', 'product_name_lenght', 'product_description_lenght', 'product_photos_qty', 'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
products = products[cols]
```
